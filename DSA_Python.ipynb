```python
# Number Hashing
n = [5, 3, 2, 2, 1, 5, 5, 5, 7, 5, 10]
m = [10, 111, 1, 9, 5, 67, 2]


freq_dict = {}

for num in n:
    freq_dict[num] = freq_dict.get(num, 0) + 1

freq_list = []

for i in m:
    if i in freq_dict:
        freq_list.append(freq_dict[i])
    else:
        freq_list.append(0)


print(freq_list)
```

# Recursion

```python
# Print multiple times
count = 0


def greet(count):
    if count == 4:
        return
    print("Meow")  # <---------- Head Recursion
    count += 1
    greet(count)


greet(count)


# ------------------------------------------------------


count = 0


def greet1(count):
    if count == 4:
        return
    count += 1
    greet1(count)
    print("Kitty")  # <---------- Tail Recursion


print()
greet1(count)
```

```python
# Recusrion using parameters

# Head Recursion


def func1(i, N):
    if i > N:
        return
    print(i)
    func1(i + 1, N)


func1(1, 7)


# Tail Recursion (Backtracking)
print()


def func1(i, N):
    if i > N:
        return
    print(i)
    func1(i + 1, N)


func1(1, 7)
```

```python
# Functional Recursion

# Sum of 1 to N
# Parameterized way


def sum_func(sum, i, N):
    if i > N:
        print(sum)
        return
    sum_func(sum + i, i + 1, N)


sum_func(0, 1, 85)


# Functional Way
def func2(N):
    if N == 1:
        return 1
    return N + func2(N - 1)


func2(15)
```

```python
# Factorial of a Number


def fact(N):
    if N == 1 or N == 0:
        return 1
    return N * fact(N - 1)


fact(5)
```

```python
# Reverse an array using Recursion
nums1 = [5, 7, 3, 2, 6, 1, 5, 9]


def rev_func(nums, left, right):
    if left >= right:
        return
    nums[left], nums[right] = nums[right], nums[left]
    rev_func(nums, left + 1, right - 1)


rev_func(nums1, 0, len(nums1) - 1)
nums1
```

```python
# check palindrome using recursion
s = "NITIN"


def palindrom_rec(s, left, right):
    if left >= right:
        return True
    if s[left] != s[right]:
        return False
    return palindrom_rec(s, left + 1, right - 1)


palindrom_rec(s, 0, len(s) - 1)
```

```python
# Fibonacci Number
def fib_rec(num):
    if num == 0 or num == 1:
        return num
    return fib_rec(num - 1) + fib_rec(num - 2)


fib_rec(8)
```

```python
# Selection Sort
nums = [5, 7, 8, 4, 1, 6, 9, 2]


def selection_sort_asc(nums):
    n = len(nums)
    for i in range(0, n):
        min_idx = i
        for j in range(i + 1, n):
            if nums[j] < nums[min_idx]:
                min_idx = j
        nums[i], nums[min_idx] = nums[min_idx], nums[i]


def selection_sort(nums):
    n = len(nums)
    for i in range(0, n):
        max_idx = i
        for j in range(i + 1, n):
            if nums[j] > nums[max_idx]:
                max_idx = j
        nums[i], nums[max_idx] = nums[max_idx], nums[i]


selection_sort(nums)
print(nums)
```

```python
# Bubble Sort (Adjacent Swap)
nums = [5, 8, 1, 6, 9, 2, 4]

n = len(nums)

for i in range(n - 2, -1, -1):
    is_swap = False
    for j in range(0, i + 1):
        if nums[j] > nums[j + 1]:
            nums[j], nums[j + 1] = nums[j + 1], nums[j]
            is_swap = True

if is_swap == False:
    print("Already Sorted")


# ---------------- OPTIMAL---------------------
```

```python
# Insertion Sort
nums = [3, 4, 5, 8, 9, 10, 7, 1]

n = len(nums)

for i in range(1, n):
    key = nums[i]
    j = i - 1
    while j >= 0 and nums[j] > key:
        nums[j + 1] = nums[j]
        j -= 1

    nums[j + 1] = key

nums
```

```python
# Merge Sort (Divide and Merge)

nums = [3, 1, 2, 4, 1, 5, 2, 6, 4]

n = len(nums)


def merge_array(left, right):
    res = []
    i, j = 0, 0
    n, m = len(left), len(right)

    while i < n and j < m:
        if left[i] <= right[j]:
            res.append(left[i])
            i += 1
        else:
            res.append(right[j])
            j += 1
    if i < n:
        while i < n:
            res.append(left[i])
            i += 1

    if j < m:
        while j < m:
            res.append(right[j])
            j += 1

    return res


def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left_half = arr[:mid]
    right_half = arr[mid:]

    left_sorted = merge_sort(left_half)
    right_sorted = merge_sort(right_half)

    return merge_array(left_sorted, right_sorted)


sr_num = merge_sort(nums)
print(sr_num)
```

```python
# Quicksort

nums = [4, 1, 7, 6, 3, 2, 8]


def partition(nums, low, high):
    pivot = nums[low]

    i = low
    j = high

    while i < j:
        while nums[i] <= pivot and i <= high - 1:
            i += 1
        while nums[j] > pivot and j >= low + 1:
            j -= 1

        if i < j:
            nums[i], nums[j] = nums[j], nums[i]

    nums[low], nums[j] = nums[j], nums[low]

    return j


def quick_sort(nums, low, high):
    if low < high:
        p_indx = partition(nums, low, high)
        quick_sort(nums, low, p_indx - 1)
        quick_sort(nums, p_indx + 1, high)


n = len(nums)
quick_sort(nums, 0, n - 1)
nums
```

```python
# 2nd largest element in array

nums = [55, 32, 97, -55, 45, 32, 88, 21]


def get_second_largest(nums):
    n = len(nums)

    largest = float("-inf")
    s_largest = float("-inf")

    for i in range(0, n):
        largest = max(largest, nums[i])

    for i in range(0, n):
        if nums[i] > s_largest and nums[i] != largest:
            s_largest = nums[i]

    return s_largest


# get_second_largest(nums)


# Optimal Solution


def get_second_largest_1(nums):
    largest = float("-inf")
    s_largest = float("-inf")

    for i in range(0, n):
        if nums[i] > largest:
            s_largest = largest
            largest = nums[i]
        elif nums[i] > s_largest and nums[i] != largest:
            s_largest = nums[i]

    return s_largest


get_second_largest_1(nums)
```

```python
# Remove duplicate from sorted array [in Place]
# return the number of unique elements

nums = [1, 1, 1, 2, 3, 4, 4, 7, 9, 9, 9, 10]


# BRUTEFORCE
freq_map = {}
n = len(nums)

for i in range(0, n):
    if n not in freq_map:
        freq_map[nums[i]] = 0

j = 0
for k in freq_map:
    nums[j] = k
    j += 1

# print(j)

# OPTIMAL


def get_num_of_unique(nums):
    n = len(nums)
    if n == 1:
        return 1

    i = 0
    j = i + 1

    while j < n:
        if nums[j] != nums[i]:
            i += 1
            nums[i], nums[j] = nums[j], nums[i]

        j += 1

    return i + 1


nums = [1, 2, 1, 1, 3, 4, 4, 7, 9, 9, 9, 10]

get_num_of_unique(nums)
```
